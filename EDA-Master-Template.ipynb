## 📌 10. Separate Numerical & Categorical Columns

```
# Numerical Columns
num_cols = df.select_dtypes(include=np.number).columns.tolist()

# Categorical Columns
cat_cols = df.select_dtypes(exclude=np.number).columns.tolist()

print("Numerical Columns:\n", num_cols)
print("\nCategorical Columns:\n", cat_cols)
```

## 📌 11. Unique Value Analysis

```
# Unique Values in Each Column
for col in df.columns:
    print(f"{col} --> {df[col].nunique()} unique values")
```

## 📌 12. Univariate Analysis - Numerical Features

```
# Histograms for Numerical Features
for col in num_cols:
    plt.figure(figsize=(8,4))
    sns.histplot(df[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.show()
```

## 📌 13. Boxplot Analysis for Numerical Features

```
# Boxplots for Numerical Features
for col in num_cols:
    plt.figure(figsize=(8,4))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

## 📌 14. Univariate Analysis - Categorical Features

```
# Countplot for Categorical Features
for col in cat_cols:
    plt.figure(figsize=(8,4))
    sns.countplot(x=df[col])
    plt.title(f"Countplot of {col}")
    plt.xticks(rotation=45)
    plt.show()
```

## 📌 15. Correlation Analysis

```
# Correlation Matrix
corr = df.corr(numeric_only=True)

plt.figure(figsize=(12,8))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

## 📌 16. Pairplot Analysis

```
# Pairplot for Numerical Features
sns.pairplot(df[num_cols])
plt.show()
```

## 📌 17. Value Counts Analysis

```
# Value Counts for Categorical Features
for col in cat_cols:
    print(f"\nValue Counts for {col}:")
    print(df[col].value_counts())
```

## 📌 18. Skewness Analysis

```
# Skewness of Numerical Features
df[num_cols].skew().sort_values(ascending=False)
```

## 📌 19. Kurtosis Analysis

```
# Kurtosis of Numerical Features
df[num_cols].kurtosis().sort_values(ascending=False)
```

## 📌 20. Outlier Detection

```
# IQR Method for Outlier Detection
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    print(f"{col} --> {outliers.shape[0]} outliers")
```

## 📌 21. Target Variable Analysis

```
# Replace 'target' with your actual target column
target_col = 'target'

print(df[target_col].value_counts())

sns.countplot(x=df[target_col])
plt.title("Target Variable Distribution")
plt.show()
```

## 📌 22. Numerical Features vs Target

```
# Numerical Features vs Target
for col in num_cols:
    if col != target_col:
        plt.figure(figsize=(8,4))
        sns.boxplot(x=df[target_col], y=df[col])
        plt.title(f"{col} vs {target_col}")
        plt.show()
```

## 📌 23. Categorical Features vs Target

```
# Categorical Features vs Target
for col in cat_cols:
    plt.figure(figsize=(8,4))
    sns.countplot(x=df[col], hue=df[target_col])
    plt.title(f"{col} vs {target_col}")
    plt.xticks(rotation=45)
    plt.show()
```

## 📌 24. Crosstab Analysis

```
# Crosstab Analysis
for col in cat_cols:
    print(f"\nCrosstab: {col} vs {target_col}")
    print(pd.crosstab(df[col], df[target_col]))
```

## 📌 25. GroupBy Analysis

```
# GroupBy Analysis
for col in cat_cols:
    print(f"\nAverage values grouped by {col}")
    print(df.groupby(col).mean(numeric_only=True))
```

## 📌 26. Feature Relationship Analysis

```
# Scatterplots for Numerical Features
for col in num_cols:
    if col != target_col:
        plt.figure(figsize=(8,4))
        sns.scatterplot(x=df[col], y=df[target_col])
        plt.title(f"{col} Relationship with {target_col}")
        plt.show()
```

## 📌 27. Distribution Analysis

```
# KDE Plots
for col in num_cols:
    plt.figure(figsize=(8,4))
    sns.kdeplot(df[col], fill=True)
    plt.title(f"KDE Plot of {col}")
    plt.show()
```
